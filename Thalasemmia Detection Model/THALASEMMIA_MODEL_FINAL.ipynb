# **THALASEMMIA CLASSIFICATION MODEL**

```python
import numpy as np
import pandas as pd
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv("thdata_final.csv")
```

```python
X = df.drop('Group', axis=1)
y = df['Group']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python

model = RandomForestClassifier(n_estimators=100, min_samples_split=2, min_samples_leaf=4,
                               max_features='sqrt', max_depth=10, random_state=42)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python

accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.2f}")
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python

joblib.dump(model, "thal_model.pkl")
print("Model saved as 'thal_model.pkl'")
```

```python

def predict_group(age, hb, mch, mchc, rdw, rbc_count):
    input_data = pd.DataFrame([[age, hb, mch, mchc, rdw, rbc_count]],
                              columns=['Age', 'Hb', 'MCH', 'MCHC', 'RDW', 'RBC count'])
    prediction = model.predict(input_data)
    return prediction[0]
```

```python
test_prediction = predict_group(age=30, hb=10.0, mch=25.0, mchc=25.0, rdw=15.0, rbc_count=5.0)
print(f"Predicted Group: {test_prediction}")
```
